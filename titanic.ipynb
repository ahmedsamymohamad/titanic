```python
import pandas as pd
```

```python
data = pd.read_csv("train_titanic.csv")
```

```python
data.head()
```

```python
df_titanic = data.drop(["PassengerId","Ticket","Cabin","Name"],axis=1)
```

```python
df_titanic.info()
```

```python
df_titanic["Age"].mean()
```

```python
df_titanic["Age"].fillna(df_titanic["Age"].mean(),inplace=True)
```

```python
df_titanic
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
df_titanic = pd.get_dummies(df_titanic,columns=['Sex','Embarked'])
```

```python
df_titanic.head()
```

```python
df_titanic['Age_Disc'] = df_titanic['Age']
scaler = MinMaxScaler()
df_titanic[['Age','Fare']] = scaler.fit_transform(df_titanic[['Age','Fare']])
```

```python
df_titanic[['Age','Fare']]
```

```python
df_titanic['FamilySize'] = df_titanic['SibSp'] + df_titanic['Parch'] +1
```

```python
df_titanic.head()
```

```python
from sklearn.feature_selection import SelectKBest, chi2
```

```python
X = df_titanic.drop('Survived',axis=1)
y = df_titanic['Survived']
selector = SelectKBest(chi2,k=4)
X_new = selector.fit_transform(X,y)
```

```python
X_new
```

```python
y
```

```python
selected_features = X.columns[selector.get_support()]
selected_features.tolist()
```

```python
bins = [0,18,30,50,80]
labels = ['Child','Young Adult','Adult','Elderly']
df_titanic['AgeGroup'] = pd.cut(df_titanic['Age_Disc'],bins=bins,labels=labels)
```
